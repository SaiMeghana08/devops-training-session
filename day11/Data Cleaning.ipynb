```python
import pandas as pd
import numpy as np
```

```python
# Create a sample dataset with data-quality problems

data = {
    "CustomerID": [101, 102, 103, 104, 105, 106, 102],
    "Name": ["Meghana", "Ravi", "Priya", "Arun", "Kiran", "Sita", "Ravi"],
    "Age": [25, np.nan, "twenty-five", -5, 40, 30, np.nan],
    "Gender": ["Female", " male ", "F", "Other", "MALE", None, " male "],
    "City": ["Chennai", " chennai ", "Bengaluru", "HYDERABAD",
             "Hyderabad", None, " chennai "],
    "Salary": ["50000", "60000", "70000", "-5000", "85,000", None, "60000"]
}

df = pd.DataFrame(data)

df.to_csv("raw_customers.csv", index=False)

df = pd.read_csv("raw_customers.csv")

print("Dataset loaded successfully!")
df
```

```python

display(df.head())

print("Shape:", df.shape)

print("\nDataset information:")
df.info()

display(df.describe(include="all"))

print("\nMissing values:")
print(df.isnull().sum())
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")


df["Salary"] = pd.to_numeric(
    df["Salary"].astype("string").str.replace(",", "", regex=False),
    errors="coerce"
)
df.loc[df["Age"] < 0, "Age"] = np.nan
df.loc[df["Salary"] < 0, "Salary"] = np.nan
df.info()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df["Gender"] = df["Gender"].fillna("Unknown")
df["City"] = df["City"].fillna("Unknown")

print(df.isnull().sum())
display(df)
```

```python
print("Duplicate rows:", df.duplicated().sum())
display(df[df.duplicated(keep=False)])
df = df.drop_duplicates()
df = df.reset_index(drop=True)
print("Shape after removing duplicates:", df.shape)
```

```python
df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip().str.title()
df["Gender"] = df["Gender"].str.strip().str.title()
df["Age"] = df["Age"].round().astype("int64")
df["Salary"] = df["Salary"].astype("float64")
df["CustomerID"] = df["CustomerID"].astype("int64")
print(df.dtypes)
display(df)
```

```python
gender_mapping = {
    "F": "Female",
    "Female": "Female",
    "M": "Male",
    "Male": "Male",
    "Other": "Other",
    "Unknown": "Unknown"
}
df["Gender"] = df["Gender"].map(gender_mapping).fillna("Unknown")
invalid_ages = df[(df["Age"] < 0) | (df["Age"] > 120)]
print("Invalid ages:")
display(invalid_ages)

invalid_salaries = df[df["Salary"] < 0]
print("Invalid salaries:")
display(invalid_salaries)

print("Gender categories:", df["Gender"].unique())
print("Cities:", df["City"].unique())
```

```python
# Transformation: create a salary band
df["SalaryBand"] = np.where(
    df["Salary"] >= 70000,
    "High",
    "Standard"
)

# Encoding: create binary columns for each gender category
df = pd.get_dummies(
    df,
    columns=["Gender"],
    prefix="Gender",
    dtype=int
)

# Normalization: scale Age and Salary to the range 0 to 1
df["Age_Normalized"] = (
    (df["Age"] - df["Age"].min()) /
    (df["Age"].max() - df["Age"].min())
    if df["Age"].max() != df["Age"].min()
    else 0.0
)

df["Salary_Normalized"] = (
    (df["Salary"] - df["Salary"].min()) /
    (df["Salary"].max() - df["Salary"].min())
    if df["Salary"].max() != df["Salary"].min()
    else 0.0
)

display(df)
```

```python
# Filter customers whose salary is at least 60000
high_salary_customers = df[df["Salary"] >= 60000]
display(high_salary_customers)

# Sort customers by salary, highest first
sorted_customers = df.sort_values(
    by="Salary",
    ascending=False
)
display(sorted_customers)

# Aggregate data by city
city_summary = df.groupby("City").agg(
    Customer_Count=("CustomerID", "count"),
    Average_Age=("Age", "mean"),
    Average_Salary=("Salary", "mean"),
    Total_Salary=("Salary", "sum")
).reset_index()

display(city_summary)
```

```python
# Export the cleaned dataset
df.to_csv("cleaned_customers.csv", index=False)

print("Cleaned dataset exported successfully!")
```
