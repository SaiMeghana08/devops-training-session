```python
import pandas as pd
```

```python
# Load the e-commerce dataset
df = pd.read_csv("ecommerce.csv")

# Display the dataset
display(df)
```

```python
# Display first 5 records
print("First 5 records:")
display(df.head())

# Display number of rows and columns
print("Dataset shape:", df.shape)

# Display column names and data types
print("\nDataset information:")
df.info()

# Display statistical summary
print("\nStatistical summary:")
display(df.describe())
```

```python
# Check whether each value is missing
print("Missing value matrix:")
display(df.isnull())

# Count missing values in every column
missing_values = df.isnull().sum()

print("\nMissing values in each column:")
print(missing_values)
```

```python
# Display only columns that contain missing values
missing_columns = missing_values[missing_values > 0]

print("Columns containing missing values:")
print(missing_columns)
```

```python
# Fill missing UnitPrice using the median
df["UnitPrice"] = df["UnitPrice"].fillna(df["UnitPrice"].median())

# Fill missing Quantity using the median
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
# Fill missing city with Unknown
df["City"] = df["City"].fillna("Unknown")

# Fill missing customer with Unknown
df["Customer"] = df["Customer"].fillna("Unknown")
```

```python
print(df.isnull().sum())
```

```python
# Count occurrences of every Order ID
order_counts = df["OrderID"].value_counts()

print("Order ID counts:")
print(order_counts)
```

```python
# Identify repeated Order IDs
duplicate_order_ids = order_counts[order_counts > 1]

print("Repeated Order IDs:")
print(duplicate_order_ids)
```

```python
# Identify repeated Order IDs
duplicate_order_ids = order_counts[order_counts > 1]

print("Repeated Order IDs:")
print(duplicate_order_ids)
```

```python
# Display all records having duplicate Order IDs
duplicate_records = df[df["OrderID"].isin(duplicate_order_ids.index)]

print("Duplicate records:")
display(duplicate_records)
```

```python
# Check for completely identical rows
print("Number of exact duplicate rows:")
print(df.duplicated().sum())
```

```python
# Remove completely duplicated rows
df = df.drop_duplicates()

# Reset index after removing duplicates
df = df.reset_index(drop=True)

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Verify that the duplicate row has been removed
print("Duplicate rows after cleaning:")
print(df.duplicated().sum())
```

```python
# Find the highest UnitPrice
highest_price = df["UnitPrice"].max()

print("Highest UnitPrice:", highest_price)
```

```python
# Find the product with the highest UnitPrice
highest_priced_product = df.loc[
    df["UnitPrice"].idxmax()
]

print("Highest-priced product:")
display(highest_priced_product)
```

```python
print(
    "Highest-priced product:",
    highest_priced_product["Product"]
)

print(
    "Price:",
    highest_priced_product["UnitPrice"]
)
```

```python
# Calculate revenue for every order
df["Revenue"] = df["UnitPrice"] * df["Quantity"]

display(df)
```

```python
# Calculate total revenue for each city
city_revenue = df.groupby("City")["Revenue"].sum()

print("City-wise revenue:")
print(city_revenue)
```

```python
# Create a city-wise revenue summary
city_revenue = (
    df.groupby("City", as_index=False)["Revenue"]
      .sum()
      .sort_values("Revenue", ascending=False)
)

display(city_revenue)
```

```python
# Fill missing numerical values
df["UnitPrice"] = df["UnitPrice"].fillna(df["UnitPrice"].median())
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())

# Fill missing city
df["City"] = df["City"].fillna("Unknown")

# Fill missing customer
df["Customer"] = df["Customer"].fillna("Unknown")

# Calculate revenue after cleaning
df["Revenue"] = df["UnitPrice"] * df["Quantity"]

display(df)
```

```python
# Calculate final city-wise revenue
city_revenue = (
    df.groupby("City", as_index=False)["Revenue"]
      .sum()
      .sort_values("Revenue", ascending=False)
)

display(city_revenue)
```
